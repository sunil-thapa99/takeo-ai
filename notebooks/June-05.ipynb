```python
'''
Compress a String Using the Counts of Repeated Characters 
Write a Python function to perform basic string compression using the counts of repeated characters.
Input: "aabcccccaaa"
Output: "a2b1c5a3"
'''

def count_rep_char(input_string: str) -> str:
    if not input_string:
        return ""

    output = ""
    count = 1

    for i in range(1, len(input_string)):
        if input_string[i] == input_string[i - 1]:
            count += 1
        else:
            output += input_string[i - 1] + str(count)
            count = 1
    output += input_string[-1] + str(count)

    return output

input_string = "aabcccccaaa"
print(count_rep_char(input_string))
```

```python
'''
Find Unique Elements from Two Lists
Question: Given two lists, find elements that are unique to each list (i.e., elements that are not shared).
Example:
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
Output: {1, 2, 5, 6}
'''
def get_unique(list1: list, list2: list) -> list:
    merge_list = list1 + list2
    counts = {}
    
    for item in merge_list:
        counts[item] = counts.get(item, 0) + 1
        
    return [i for i in merge_list if counts[i] == 1]

list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
print(get_unique(list1, list2))
```

```python
def get_unique(list1: list, list2: list) -> list:
    list1_set, list2_set = set(list1), set(list2)
    unique_elem = []

    for i in list1:
        if i not in list2_set:
            unique_elem.append(i)
    
    for i in list2:
        if i not in list1_set:
            unique_elem.append(i)

    return unique_elem

list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
print(get_unique(list1, list2))
```

```python
'''
Find Missing Number in an Array
Question: Given a list of integers from 1 to `n` with one number missing, find the missing number using sets.
Example:
nums = [1, 2, 4, 5]
Output: 3
'''

def find_missing(nums: list) -> int:
    nums_set = set(nums)
    for i in range(1, len(nums)+2):
        if i not in nums_set:
            return i
        
    return -1

nums = [1, 2, 4, 5]
print(find_missing(nums))
```
