```python
'''
Check for Anagrams
Write a Python function to check if two given strings are anagrams of each other.
Input: "listen", "silent"
Output: True
'''
def check_anagram(first_str: str, second_str: str) -> bool:
    first_str, second_str = sorted(first_str.lower()), sorted(second_str.lower())
    if first_str == second_str:
        return True
    return False

first_inp = 'listen'
second_inp = 'silent'
print(check_anagram(first_inp, second_inp))
```

```python
'''
Count Vowels in a String
Write a Python function to count the number of vowels (a, e, i, o, u) in a given string.
Input: "Hello World"
Output: 3
'''
vowel = ['a', 'e', 'i', 'o', 'u']
def count_vowel(word: str) -> int:
    count = 0
    for w in word:
        if w.lower() in vowel:
            count += 1
    return count

word = 'Hello World'
print(count_vowel(word))
```

```python
'''
Remove Duplicate Characters
Write a Python function to remove all duplicate characters in a given string while preserving the order of the remaining characters.
Input: "programming"
Output: "progamin"
'''

def rem_dup(word: str) -> str:
    return_word = ''
    for w in word:
        if w not in return_word:
            return_word += w
    return return_word

input_word = "programming"
print(rem_dup(input_word))
```

```python
'''
Find the First Non-Repeating Character
Write a Python function to find the first non-repeating character in a given string and return its index.
Input: "swiss"
Output: 1 (for 'w' in "swiss")
'''

def first_non_rep_char(word: str) -> int:
    for i, w in enumerate(word):
        if word.count(w) == 1:
            return i
    return -1
input_word = 'swiss'
print(first_non_rep_char(input_word))
input_word = 'abbbacc'
print(first_non_rep_char(input_word))
input_word = 'aabbsssc'
print(first_non_rep_char(input_word))
```
