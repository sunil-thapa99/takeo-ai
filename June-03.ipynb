```python
'''
Write a Python function to flatten a nested list.
Input: [[1, 2], [3, 4], [5]]
Output: [1, 2, 3, 4, 5]
'''
def flatten(input_list: list) -> list:
    return [i for inn_list in input_list for i in inn_list]

input_list = [[1, 2], [3, 4], [5]]
print(flatten(input_list))
```

```python
'''
Write a Python function to find the second largest element in a list.
Input: [1, 2, 3, 4, 5]
Output: 4
'''
def sec_larg(input_list: list) -> list:
    return sorted(input_list, reverse=True)[1]
input_list = [1, 2, 3, 4, 5]
print(sec_larg(input_list))
```

```python
'''
Write a Python function to find all pairs in a list that sum to a specific value.
Input: [1, 2, 3, 4, 5], Sum=6
Output: [(1, 5), (2, 4)]
'''
def sum_pair(input_list: list, sum: int) -> list:
    seen = set()
    pairs = []

    for i in input_list:
        remainder = sum - i
        if remainder in seen:
            pairs.append((i, remainder))
        seen.add(i)
    return pairs

input_list = [1, 2, 3, 4, 5]
sum = 6
print(sum_pair(input_list, sum))
```
